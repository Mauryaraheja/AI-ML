Now Working with Numerical Data

a. Histogram

```python
import matplotlib.pyplot as plt
```

```python
import numpy as np

df['price_in_lakhs'] = np.where(
    df['price_unit'] == 'Cr',
    df['price'] * 100,   # 1 Cr = 100 L
    df['price']          # already in L
)

plt.hist(df['price_in_lakhs'], bins=50, range=(0, 2500))
plt.xlabel('Price (in Lakhs)')
plt.show()
df['price_in_lakhs'].skew()

#skewness here tells the shape of our distribution , more positive the skew value is more the graph is right skewed , and if skew value in negative , it indicates graph is left skewed
```

```python
g = sns.displot(df['price_in_lakhs'], kde=True, stat='density', bins=50, binrange=(0, 1000))
g.set(xlim=(0, 1000))
#here density - otherwise y axis will show number of counts , here it is showing probability using density
#here binrange effects graph , while xlim = 0,2500 only shows that part which we want but doesn't effect graph
#Here kde is kernel density estimation
```

Distplot is going to be removed in the upcoming updates of seaborn , that's why i used displot method , written above
