# **HOW TO EXTRACT DATA THROUGH API**

```python
import pandas as pd
import requests
#in this we extracted a api request from a movies website , through provided api key
#We will extract only those columns which are useful for us , we will not extract unnecessary ones
```

```python
response = requests.get('https://api.themoviedb.org/3/movie/550/recommendations?language=en-US&page=1&api_key=683c572d766157aedff1dc7d2890ea7a')
```

```python
df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_count']]
```

in this by .json we converted the data in frames by using dataframes , and the list inside list that we used was for the columns that we need

```python
df = pd.DataFrame()
```

```python
#created a empty dataframe , as number of pages are 17 , and we only fetched data of page 1 , we need a loop to run through 17 pages
```

```python
for i in range(1,18):
  response = requests.get('https://api.themoviedb.org/3/movie/550/recommendations?language=en-US&page={}&api_key=683c572d766157aedff1dc7d2890ea7a'.format(i))
  temp_df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_count']]
  df = pd.concat([df,temp_df], ignore_index = True)
```
