This becomes hectic if we see heading by heading , or tags by tags , so it's better to see data for each book , that is correct method below

```python
books =soup.find_all('article', class_='product_pod')
```

```python
name = []
rating = []
price_color = []

for i in books :
  name.append(i.find('h3').text.strip())
  rating.append(i.find('p',class_ ='star-rating')['class'][1])
  price_color.append(i.find('p',class_='price_color').text.strip())



d = {'name':name , 'rating':rating , 'price_color':price_color }
df = pd.DataFrame(d)

#this class means class = star-rating three for ex, so class has star-rating and three as different indexes , 1 chooses second one that is three

#if i write name.append(print(i.find('h3').text.strip())) - it will return nothing it will just print the values so the title name will not get stored inside the list
```

```python
final = pd.DataFrame()

for j in range (1,51):
  url = 'https://books.toscrape.com/catalogue/page-{}.html'.format(j)

  headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) AppleWebKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
  response  = requests.get(url,headers = headers)
  response.encoding = "utf-8"

  webpage = response.text
  soup = BeautifulSoup(webpage, "html.parser")
  soup = BeautifulSoup(webpage ,'lxml')

  books = soup.find_all('article', class_='product_pod')

  name = []
  rating = []
  price_color = []

  for i in books :
    name.append(i.find('h3').text.strip())
    rating.append(i.find('p',class_ ='star-rating')['class'][1])
    price_color.append(i.find('p',class_='price_color').text.strip())



  d = {'name':name , 'rating':rating , 'price_color':price_color }
  df = pd.DataFrame(d)

  final = pd.concat([final,df],ignore_index=True)

```
